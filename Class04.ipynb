# **Elementary Functions: Exponential, Logarithmic, Trigonometric & Hyperbolic**
**Elementary functions** comprise a core set of mathematical tools used to model diverse real-world phenomena, grouped into four major classes based on their geometric and algebraic origins. **Exponential functions** (like $e^x$) feature a variable in the exponent, representing rapid growth or decay, while **logarithmic functions** (like $\ln(x)$) act as their direct inverses, scaling down vast data ranges. **Trigonometric functions** (like $\sin(x)$ and $\cos(x)$) are rooted in the geometry of the unit circle and are inherently periodic, making them ideal for modeling wave-like, repeating behaviors. In contrast, **hyperbolic functions** (like $\sinh(x)$ and $\cosh(x)$) are built from exponential terms and relate to the geometry of a hyperbola, uniquely describing non-periodic physical structures such as the natural curve of a hanging power line.

## **Example: Functions**
Here is a complete Python example using matplotlib and numpy to visualize and compare all four classes of elementary functions (Exponential, Logarithmic, Trigonometric and Hyperbolic) side-by-side.

```python
# Importing all the necessary libraries and resources:
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Setting up the data ranges:
x_exp_hyper = np.linspace(-2, 2, 400)      # Range for Exponential and Hyperbolic
x_log = np.linspace(0.05, 5, 400)          # Range for Logarithmic (must be > 0)
x_trig = np.linspace(-2 * np.pi, 2 * np.pi, 400)  # Range for Trigonometric (full cycles)
```

```python
# Creating the plots:
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Visualizing Elementary Functions', fontsize=16, fontweight='bold')

# Top Left: Exponential:
axs[0, 0].plot(x_exp_hyper, np.exp(x_exp_hyper), label=r'$e^x$ (Growth)', color='firebrick', lw=2)
axs[0, 0].plot(x_exp_hyper, np.exp(-x_exp_hyper), label=r'$e^{-x}$ (Decay)', color='salmon', lw=1.5, linestyle='--')
axs[0, 0].set_title('Exponential Functions')
axs[0, 0].grid(True, alpha=0.3)
axs[0, 0].legend()

# Top Right: Logarithmic:
axs[0, 1].plot(x_log, np.log(x_log), label=r'$\ln(x)$', color='darkgreen', lw=2)
axs[0, 1].axvline(0, color='black', linestyle=':', alpha=0.5) # Vertical asymptote
axs[0, 1].set_title('Logarithmic Functions')
axs[0, 1].grid(True, alpha=0.3)
axs[0, 1].legend()

# Bottom Left: Trigonometric:
axs[1, 0].plot(x_trig, np.sin(x_trig), label=r'$\sin(x)$', color='royalblue', lw=2)
axs[1, 0].plot(x_trig, np.cos(x_trig), label=r'$\cos(x)$', color='skyblue', lw=1.5, linestyle='--')
axs[1, 0].set_title('Trigonometric Functions (Periodic)')
axs[1, 0].grid(True, alpha=0.3)
axs[1, 0].legend()

# Bottom Right: Hyperbolic:
axs[1, 1].plot(x_exp_hyper, np.sinh(x_exp_hyper), label=r'$\sinh(x)$', color='purple', lw=2)
axs[1, 1].plot(x_exp_hyper, np.cosh(x_exp_hyper), label=r'$\cosh(x)$', color='violet', lw=2)
axs[1, 1].set_title
```
