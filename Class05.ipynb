# **Limits & Lateral Limits**
In calculus, a **limit** describes the value that a function approaches as the input gets closer and closer to a specific point, regardless of whether the function actually reaches or is defined at that exact point. **Lateral limits** (or one-sided limits) break this concept down by approaching the target point from only one direction: the **left-hand limit** examines the function's behavior as x approaches from smaller values ($x \to c^-$), while the **right-hand limit** approaches from larger values ($x \to c^+$). For a standard, two-sided limit to exist at a point, both lateral limits must exist and be equal to the exact same value, mathematically expressed as:

$$\lim_{x \to c^-} f(x) = \lim_{x \to c^+} f(x) = L$$

```python
# Importing all tne necessary libraries:
import sympy as sp
```

```python
# Defining the variable and the function:
x = sp.Symbol('x')

# Creating a piecewise function that has a jump at x = 2
f = sp.Piecewise((x**2, x < 2), (x + 5, x >= 2))
```

```python
# Left-hand limit as x approaches 2 from the negative side (-):
left_limit = sp.limit(f, x, 2, dir='-')

# Displaying the result:
print(f'Left-hand limit (x -> 2-): {left_limit}')    # Outputs: 4 (since 2^2 = 4)
```

```python
# Right-hand limit as x approaches 2 from the positive side (+):
right_limit = sp.limit(f, x, 2, dir='+')

# Displaying the result:
print(f'Right-hand limit (x -> 2+): {right_limit}')  # Outputs: 7 (since 2 + 5 = 7)
```

```python
# Two-sided limit:
two_sided_limit = sp.limit(f, x, 2)

# Displaying the result:
print(f'Two-sided limit (x -> 2): {two_sided_limit}') # Outputs: dir='+-' is undefined or raises an error because 4 != 7
```
