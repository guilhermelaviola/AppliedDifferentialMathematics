# **Practical Activity**

For this exercise, let's consider two functions f and g. For both functions, let's also consider an element x that is in the domain of f and g. In this case, we can calculate:
- *f+g*
- *f-g*
- *f⋅g*
- *f/g*, with *g(x) ≠ 0*

For ***f(x)=2x+4f(x)=2x+4f(x)=2x+4​​ and g(x)=6g(x)=6g(x)=6​***​ functions of one variable x belonging to the set of real numbers, present, either by formula or graph, the function resulting from the four operations above.

```python
# Importing all the necessary libraries and resouces:
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Defining the input domain (x values from -5 to 5):
x = np.linspace(-5, 5, 400)

# Defining the original functions f(x) and g(x):
f = 2 * x + 4
g = np.full_like(x, 6)  # Creates an array of 6s with the same shape as x

# Performing the four operations:
sum_fg = f + g       # Addition
diff_fg = f - g      # Subtraction
prod_fg = f * g      # Multiplication
div_fg = f / g       # Division

# Creating a 2x2 grid of subplots for visualization:
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Plotting Addition:
axs[0, 0].plot(x, sum_fg, label='$f(x) + g(x) = 2x + 10$', color='blue')
axs[0, 0].set_title('Addition: $(f+g)(x)$')
axs[0, 0].grid(True)
axs[0, 0].axhline(0, color='black', linewidth=0.5)
axs[0, 0].axvline(0, color='black', linewidth=0.5)
axs[0, 0].legend()

# Plotting Subtraction:
axs[0, 1].plot(x, diff_fg, label='$f(x) - g(x) = 2x - 2$', color='orange')
axs[0, 1].set_title('Subtraction: $(f-g)(x)$')
axs[0, 1].grid(True)
axs[0, 1].axhline(0, color='black', linewidth=0.5)
axs[0, 1].axvline(0, color='black', linewidth=0.5)
axs[0, 1].legend()

# Plotting Multiplication:
axs[1, 0].plot(x, prod_fg, label='$f(x) \cdot g(x) = 12x + 24$', color='green')
axs[1, 0].set_title('Multiplication: $(f \cdot g)(x)$')
axs[1, 0].grid(True)
axs[1, 0].axhline(0, color='black', linewidth=0.5)
axs[1, 0].axvline(0, color='black', linewidth=0.5)
axs[1, 0].legend()

# Plotting Division:
axs[1, 1].plot(x, div_fg, label='$\\frac{f(x)}{g(x)} = \\frac{1}{3}x + \\frac{2}{3}$', color='red')
axs[1, 1].set_title('Division: $(f/g)(x)$')
axs[1, 1].grid(True)
axs[1, 1].axhline(0, color='black', linewidth=0.5)
axs[1, 1].axvline(0, color='black', linewidth=0.5)
axs[1, 1].legend()

# Adjusting the layout and saving the visualization as PNG:
plt.tight_layout()
plt.savefig('operations_plot.png')
```
