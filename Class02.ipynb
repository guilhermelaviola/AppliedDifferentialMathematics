# **Functions: Definition, Graph, Operations & Common Types**
A mathematical function is a strict rule that maps each unique input from a set called the domain to exactly one output in the range, a property that can be visually verified using the vertical line test on its graph. These functions can be combined through arithmetic operations like addition, subtraction, multiplication, and division, or nested within one another through a process called composition ($f(g(x))$). Depending on their algebraic rules, functions exhibit distinct geometric behaviors and are categorized into common types, including straight-line linear functions, parabolic quadratic functions, multi-curved polynomials, asymptotic rational functions, and rapidly changing exponential or logarithmic functions.

## **Example: Simple Functions**
The following example defines two simple functions ($f(x)$ and $g(x)$). It performs a function composition, and then plots the results.

```python
# Importing all the necessary libraries and resources:
import matplotlib.pyplot as plt
import numpy as np

# Defining the functions:
def f(x):
    '''A quadratic function: f(x) = x^2'''
    return x**2

def g(x):
    '''A linear function: g(x) = 2x + 3'''
    return 2 * x + 3

def fg_composite(x):
    '''Function Composition: f(g(x)) = (2x + 3)^2'''
    return f(g(x))

# Generating input values (Domain):
x_values = np.linspace(-5, 5, 400)

# Calculating output values (Range):
y_f = f(x_values)
y_g = g(x_values)
y_composite = fg_composite(x_values)

# Plotting the graphs:
plt.figure(figsize=(10, 6))

plt.plot(x_values, y_f, label='f(x) = x² (Quadratic)', color='blue', linewidth=2)
plt.plot(x_values, y_g, label='g(x) = 2x + 3 (Linear)', color='green', linewidth=2)
plt.plot(
    x_values,
    y_composite,
    label='f(g(x)) = (2x + 3)² (Composite)',
    color='purple',
    linestyle='--',
    linewidth=2,
)

# Formatting the graph:
plt.axhline(0, color='black', linewidth=0.5, linestyle=':')
plt.axvline(0, color='black', linewidth=0.5, linestyle=':')
plt.title('Visualizing Functions and Composition in Python')
plt.xlabel('Domain (x)')
plt.ylabel('Range (y)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(-5, 30)

# Showing the plot:
plt.show()
```
