# **Important Functions: Even, Odd, Injective, Surjective & Bijective**
Mathematical functions can be categorized by their symmetry or how they map elements between sets: **even** functions are symmetrical across the y-axis because $f(-x) = f(x)$, while **odd** functions exhibit 180-degree rotational symmetry around the origin because $f(-x) = -f(x)$. When analyzing mappings from a domain to a codomain, a function is **injective** (one-to-one) if every distinct input maps to a unique output, **surjective** (onto) if its range completely covers the entire codomain, and **bijective** if it satisfies both conditions simultaneously, creating a perfect, invertible one-to-one pairing between both sets.

## **Example: Function Properties**
The following example implements and tests examples for each of the function properties (Injective, Surjective and Bijective), using standard mathematical operations.

```python
# Defining the functions:
def is_even(f, domain):
    '''Checks if a function is even: f(-x) == f(x)'''
    return all(f(-x) == f(x) for x in domain)

def is_odd(f, domain):
    '''Checks if a function is odd: f(-x) == -f(x)'''
    return all(f(-x) == -f(x) for x in domain)

def analyze_mapping(f, domain, codomain):
    '''
    Analyzes a function's mapping properties (Injective, Surjective, Bijective)
    based on a specific domain and codomain.
    '''
    # Calculating all outputs (the range):
    outputs = [f(x) for x in domain]

    # Injective: Distinct inputs must give distinct outputs:
    is_injective = len(set(outputs)) == len(domain)

    # Surjective: Every element in the codomain must be an output:
    is_surjective = set(outputs) == set(codomain)

    # Bijective: Both injective and surjective:
    is_bijective = is_injective and is_surjective

    return {
        'Injective (One-to-One)': is_injective,
        'Surjective (Onto)': is_surjective,
        'Bijective': is_bijective
    }
```

```python
# Testing the Functions

# Parity Tests (Even/Odd):
test_domain = [-3, -2, -1, 0, 1, 2, 3]

f_even = lambda x: x**2
f_odd = lambda x: x**3

print('--- Parity Checks ---')
print(f'Is x^2 even? {is_even(f_even, test_domain)}')
print(f'Is x^3 odd? {is_odd(f_odd, test_domain)}')
print()

# Mapping Tests (Injective, Surjective, Bijective):
domain_set = [1, 2, 3]
codomain_set = [2, 4, 6]

# This function perfectly doubles each input:
f_bijective = lambda x: x * 2

print('--- Mapping Checks ---')
print(f'Properties for f(x) = 2x with Domain {domain_set} and Codomain {codomain_set}:')
results = analyze_mapping(f_bijective, domain_set, codomain_set)
for property_name, status in results.items():
    print(f' - {property_name}: {status}')
```
