# **Elementary Derivatives: Exponential, Logarithmic, Trigonometric & Hyperbolic**
**Elementary derivatives** represent the fundamental rules used to calculate the instantaneous rate of change for core mathematical functions. For **exponential functions**, the derivative is proportional to the function itself (most notably, $\frac{d}{dx}(e^x) = e^x$), while **logarithmic functions** derive into reciprocal forms, where $\frac{d}{dx}(\ln x) = \frac{1}{x}$. **Trigonometric derivatives** describe the rates of oscillating geometric wave functions, cycling through predictable patterns like $\frac{d}{dx}(\sin x) = \cos x$ and $\frac{d}{dx}(\cos x) = -\sin x$. Similarly, **hyperbolic derivatives** govern the calculus of curves shaped by hanging chains or cables, mapping functions to their hyperbolic counterparts without the sign changes seen in trigonometry, such as $\frac{d}{dx}(\sinh x) = \cosh x$ and $\frac{d}{dx}(\cosh x) = \sinh x$.

## **Example: Derivatives**
Following are examples using SymPy, Python's premier symbolic mathematics library. It allows you to define mathematical symbols and calculate exact derivatives just like you would on paper.

```python
# Importing all the necessary libraries:
import sympy as sp

# Defining the symbolic variable:
x = sp.Symbol('x')
```

```python
# Exponential Functions: f(x) = e^x and g(x) = 2^x:
f_exp1 = sp.exp(x)
f_exp2 = 2**x

df_exp1 = sp.diff(f_exp1, x)
df_exp2 = sp.diff(f_exp2, x)
```

```python
# Logarithmic Functions: f(x) = ln(x) and g(x) = log_10(x):
f_log1 = sp.log(x)  # sp.log is the natural logarithm (ln)
f_log2 = sp.log(x, 10)

df_log1 = sp.diff(f_log1, x)
df_log2 = sp.diff(f_log2, x)
```

```python
# Trigonometric Functions: f(x) = sin(x) and g(x) = tan(x):
f_trig1 = sp.sin(x)
f_trig2 = sp.tan(x)

df_trig1 = sp.diff(f_trig1, x)
df_trig2 = sp.diff(f_trig2, x)
```

```python
# Hyperbolic Functions: f(x) = sinh(x) and g(x) = cosh(x):
f_hyp1 = sp.sinh(x)
f_hyp2 = sp.cosh(x)

df_hyp1 = sp.diff(f_hyp1, x)
df_hyp2 = sp.diff(f_hyp2, x)
```

```python
# Displaying the Results:
print('Elementary Derivatives\n')
print(f'Exponential:\n   d/dx({f_exp1}) = {df_exp1}\n   d/dx({f_exp2}) = {df_exp2}\n')
print(f'Logarithmic:\n   d/dx({f_log1}) = {df_log1}\n   d/dx({f_log2}) = {df_log2}\n')
print(f'Trigonometric:\n   d/dx({f_trig1}) = {df_trig1}\n   d/dx({f_trig2}) = {df_trig2}\n')
print(f'Hyperbolic:\n   d/dx({f_hyp1}) = {df_hyp1}\n   d/dx({f_hyp2}) = {df_hyp2}')
```
