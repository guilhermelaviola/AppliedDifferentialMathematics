# **Indeterminate Limits & Limits at Infinity**
**Indeterminate limits** occur when evaluating a limit yields an unspecified form like $0/0$ or $\infty/\infty$, meaning the direct substitution method fails and further algebraic manipulation—such as factoring, rationalizing, or applying L'Hôpital's Rule—is required to find the actual limit value. In contrast, **limits at infinity** describe the behavior of a function as the independent variable $x$ grows without bound toward positive or negative infinity ($x \to \infty$ or $x \to -\infty$), which effectively helps determine the end behavior of the function and identifies whether it approaches a finite value, known as a horizontal asymptote, or diverges entirely.

## **Example: Limits**
The following example uses the sympy library (a powerful symbolic mathematics package) to define and calculate both an indeterminate limit and a limit at infinity.

```python
# Importing all the necessary libraries and resources:
import sympy as sp
```

```python
# Defining the symbolic variable:
x = sp.Symbol('x')
```

```python
# Indeterminate Limit example:
# Defining a function that yields a 0/0 indeterminate form at x = 2. Direct substitution: (2^2 - 4) / (2 - 2) = 0/0:
f_indeterminate = (x**2 - 4) / (x - 2)

# Calculating the limit as x approaches 2:
limit_ind = sp.limit(f_indeterminate, x, 2)

print(f'Function: f(x) = (x^2 - 4) / (x - 2)')
print(f'Limit as x -> 2: {limit_ind}\n')
```

```python
# Limit at Infinity example:
# Defining a function to test end behavior as x goes to infinity. This function has a horizontal asymptote:
f_infinity = (3 * x**2 + 5) / (2 * x**2 - x)

# Calculating the limit as x approaches positive infinity:
limit_inf = sp.limit(f_infinity, x, sp.oo)

print(f'Function: g(x) = (3x^2 + 5) / (2x^2 - x)')
print(f'Limit as x -> oo (infinity): {limit_inf}')
```
