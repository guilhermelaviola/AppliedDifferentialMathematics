# **Maxima & Minima**
In mathematics, the maxima and minima (collectively known as extrema) refer to the largest and smallest values that a function takes, either within a specific neighborhood or across its entire domain. A local maximum or local minimum occurs at a point where the function's value is higher or lower than all immediately surrounding points, typically found where the derivative of the function equals zero ($f'(x) = 0$) or is undefined, marking a peak or a valley on a graph. In contrast, the global (or absolute) maximum and global minimum represent the absolute highest and lowest outputs achieved by the function over its entire allowable domain.

## **Example: Maxima & Minima**
The following example demonstrates how to find the local and global maxima and minima of a function using the scipy.optimize library.

```python
# Importing all the necessary libraries and resources:
import numpy as np
from scipy.optimize import minimize_scalar
```

```python
# Defining a function: f(x) = x^4 - 4x^2 + x. This function has multiple peaks and valleys (local and global extrema):
def f(x):
    return x**4 - 4*x**2 + x
```

```python
# Finding the Global Minimum. Since the function goes to infinity at the edges, we look within a reasonable bounds:
global_min_res = minimize_scalar(f, bounds=(-3, 3), method='bounded')
print(f'Global Minimum found at x = {global_min_res.x:.4f}, f(x) = {global_min_res.fun:.4f}')
```

```python
# Finding a Local Minimum. By changing the bounds, we can target the other 'valley':
local_min_res = minimize_scalar(f, bounds=(0, 2), method='bounded')
print(f'Local Minimum found at x = {local_min_res.x:.4f}, f(x) = {local_min_res.fun:.4f}')
```

```python
# Finding the Local Maximum. To find a maximum using a minimization tool, we minimize the negative of the function: -f(x):
def neg_f(x):
    return -f(x)

local_max_res = minimize_scalar(neg_f, bounds=(-1, 1), method='bounded')
print(f'Local Maximum found at x = {local_max_res.x:.4f}, f(x) = {-local_max_res.fun:.4f}')
```
