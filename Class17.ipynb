# **Taylor's Formula**
**Taylor's Formula** (or Taylor's Theorem) provides a way to approximate a sufficiently differentiable function near a specific point using a polynomial. It states that if a function $f(x)$ is $n+1$ times differentiable on an interval containing $a$, then the function can be expressed as the sum of a Taylor polynomial of degree $n$ and a remainder term $R_n(x)$ that accounts for the approximation error:

$$f(x) = f(a) + f'(a)(x-a) + \frac{f''(a)}{2!}(x-a)^2 + \dots + \frac{f^{(n)}(a)}{n!}(x-a)^n + R_n(x)$$

This formula is fundamental in calculus and numerical analysis because it allows complex, non-linear functions to be broken down into simpler polynomial terms, making them much easier to analyze, integrate, or compute computationally near the point of approximation.

## **Example: Taylor's Formula**
The following example uses the math function $f(x) = e^x$ approximated around the point $a = 0$ (which is also known as a Maclaurin series). Here we use Taylor's Formula to build a 3rd-degree polynomial to approximate $e^{0.5}$ and compare it to Python's built-in, highly accurate value.

```python
# Importing all the necessary libraries and resources:
import math
```

```python
# Approximates e^x around point 'a' using a Taylor polynomial up to the specified degree.
# For f(x) = e^x, all derivatives f'(x), f''(x), etc., are also e^x.
def taylor_exponential_approx(x, a, degree):
  approximation = 0.0
  for n in range(degree + 1):
    # Calculating the n-th derivative at point 'a'
    # Since f(x) = e^x, the n-th derivative at 'a' is always e^a:
    nth_derivative_at_a = math.exp(a)

    # Calculating the term: [f^(n)(a) / n!] * (x - a)^n:
    term = (nth_derivative_at_a / math.factorial(n)) * ((x - a) ** n)

    # Adding the term to our polynomial sum:
    approximation += term

    return approximation
```

```python
# Parameters for the example above:
target_x = 0.5   # The point we want to estimate (e^0.5)
base_a = 0.0     # The point we are approximating near (a = 0)
```

```python
# True mathematical value:
true_value = math.exp(target_x)

print(f"True value of e^{target_x}: {true_value}\n")
print("Taylor Formula Approximations:")
print("-" * 45)
```

```python
# Calculating approximations for degrees 1, 2, and 3 to see it get more accurate:
for deg in range(1, 4):
    approx = taylor_exponential_approx(target_x, base_a, deg)
    error = abs(true_value - approx) # This represents the remainder term R_n(x)
    print(f"Degree {deg} Polynomial: {approx:.6f} (Error: {error:.6f})")
```
