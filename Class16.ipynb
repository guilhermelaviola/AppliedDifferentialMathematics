# **L'Hospital's Rule**
**L'Hôpital's Rule** is a calculus theorem used to evaluate the limits of indeterminate forms—specifically when direct substitution yields $\frac{0}{0}$ or $\frac{\infty}{\infty}$. It states that if two functions, $f(x)$ and $g(x)$, are differentiable near a point $c$ (except possibly at $c$) and their limits as $x$ approaches $c$ both equal zero or both approach infinity, then the limit of their quotient is equal to the limit of the quotient of their derivatives, expressed as $\lim_{x \to c} \frac{f(x)}{g(x)} = \lim_{x \to c} \frac{f'(x)}{g'(x)}$, provided the limit on the right exists or is infinite and $g'(x) \neq 0$ near $c$.

## **Example: L'Hôpital's Rule**
The follwing example uses the sympy library, which is perfect for symbolic mathematics and calculus. This script defines a classic indeterminate limit, $\lim_{x \to 0} \frac{\sin(x)}{x}$, and demonstrates both the standard limit evaluation and the manual application of L'Hôpital's Rule by differentiating the numerator and denominator.Python

```python
# Importing all the necessary libraries:
import sympy as sp
```

```python
# Defining the symbolic variable:
x = sp.Symbol('x')

# Defining the numerator and denominator functions:
f = sp.sin(x)
g = x
```

```python
print("Checking the Indeterminate Form")
# Direct substitution yields 0/0
print(f"Direct substitution at x=0: f(0)/g(0) = {f.subs(x, 0)}/{g.subs(x, 0)}")

print("\nApplying L'Hôpital's Rule")
# Taking the derivatives of the numerator and denominator:
f_prime = sp.diff(f, x)
g_prime = sp.diff(g, x)

print(f"Derivative of numerator f'(x): {f_prime}")
print(f"Derivative of denominator g'(x): {g_prime}")
```

```python
# Evaluating the limit of the quotient of the derivatives:
limit_lhospital = sp.limit(f_prime / g_prime, x, 0)
print(f"Limit of f'(x)/g'(x) as x -> 0: {limit_lhospital}")

print("\nVerification")
# Verifying with SymPy's built-in limit function:
standard_limit = sp.limit(f / g, x, 0)
print(f"Standard limit of f(x)/g(x) as x -> 0: {standard_limit}")
```
