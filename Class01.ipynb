# **Number Sets, Inequalities & Absolute Value**
In Applied Differential Mathematics, foundational algebraic concepts like number sets, inequalities, and absolute values cease to be just abstract rules. Instead, they become essential tools used to define the boundaries, behaviors, and constraints of dynamic, changing systems (like fluid flow, heat transfer, or economic growth).

## **Example: Example: Continuous Intervals and Complex Roots**
In applied math, we often need to represent a continuous interval of real numbers ($\mathbb{R}$) or work with complex numbers ($\mathbb{C}$). Python handles these using native types and NumPy.

```python
import numpy as np

# Representing a continuous Real Number Set [0, 10] for time (t)
# Genenerating 100 points between 0 and 10 to simulate continuous time:
time_set = np.linspace(0, 10, 100)
print(f'Real Number Interval (First 5 points): {time_set[:5]}')

# Representing Complex Numbers (C):
# It's common when solving differential equations for oscillating systems (like a pendulum):
root1 = complex(2, 3)   # 2 + 3j
root2 = 2 - 3j          # Alternative syntax

print(f'Complex Root 1: {root1}')
print(f'Real part: {root1.real}, Imaginary part: {root1.imag}')
```

## **Example: Simulating a Chemical Tank with a Maximum Volume Constraint**
Inequalities are used to define boundary conditions, safety limits, or to check for system stability.

```python
import matplotlib.pyplot as plt

# Parameters:
dt = 0.1            # Time step
t_max = 5.0         # Total time
V = 20.0            # Initial volume in liters
flow_rate = 15.0    # Liters per second pouring in

# Inequality Constraints:
V_max = 50.0        # Physical limit of the tank (V <= V_max)
V_min = 0.0         # (V >= 0)

time_steps = [0]
volume_steps = [V]

# Differential simulation loop:
for i in range(int(t_max / dt)):
    # dV/dt = flow_rate -> V_next = V + flow_rate * dt
    V_next = V + flow_rate * dt

    # Applying the inequality constraint: Volume cannot exceed V_max:
    if V_next >= V_max:
        V = V_max
    elif V_next <= V_min:
        V = V_min
    else:
        V = V_next

    time_steps.append(time_steps[-1] + dt)
    volume_steps.append(V)

# Plotting the constrained system:
plt.plot(time_steps, volume_steps, label='Tank Volume')
plt.axhline(y=V_max, color='r', linestyle='--', label='V_max Constraint')
plt.xlabel('Time (s)')
plt.ylabel('Volume (L)')
plt.legend()
plt.title('Differential System with Inequality Constraints')
plt.show()
```

## **Example: Error Bound Checking in Numerical Integration**
Absolute value is widely used to calculate error margins, step-size adjustments, or to model non-smooth events where a derivative's sign changes abruptly.

```python
import numpy as np

# Suppose the exact analytical solution of a differential equation at t=2 is e^2:
exact_solution = np.exp(2)

# Your numerical approximation (e.g., from Euler's Method):
numerical_approx = 7.305

# Define acceptable error tolerance (Inequality + Absolute Value):
tolerance = 0.1

# Calculating absolute error: |Exact - Approx|:
absolute_error = abs(exact_solution - numerical_approx)

print(f'Exact: {exact_solution:.4f}')
print(f'Approx: {numerical_approx:.4f}')
print(f'Absolute Error: {absolute_error:.4f}')

# Checking if the error satisfies the bound constraint:
if absolute_error <= tolerance:
    print('Success: The approximation is within the acceptable error bound.')
else:
    print('Warning: Error exceeds tolerance. Reduce step size (h).')
```
