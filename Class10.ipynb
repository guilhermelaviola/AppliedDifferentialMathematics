# **Differentiation Rules & Composite/Inverse**
Differentiation rules are fundamental algebraic formulas—such as the power, product, and quotient rules—used to systematically compute the derivative (or rate of change) of standard functions without relying on the tedious limit definition. When dealing with more complex, layered functions, these rules extend to composite functions via the Chain Rule, which differentiates the "outer" function while leaving the "inner" function intact, then multiplies it by the derivative of that inner function ($$(f \circ g)'(x) = f'(g(x)) \cdot g'(x)$$

## **Example: Standard Differentiation Rules**
The following example uses the sympy library (a powerful symbolic mathematics library) to demonstrate standard differentiation rules, composite functions (Chain Rule), and inverse functions exactly as described.

```python
# Importing all the necessary libraries and resources:
import sympy as sp

# Defining the symbolic variable:
x = sp.Symbol('x')
```

```python
# Standard Differentiation Rules (Power & Product Rule): f(x) = x^3 * sin(x):
f = x**3 * sp.sin(x)
df_dx = sp.diff(f, x)
print("Standard Rules Example:")
print(f"Function: f(x) = {f}")
print(f"Derivative: f'(x) = {df_dx}\n")
```

```python
# Composite Function & Chain Rule: g(x) = ln(x^2 + 1) -> Outer function is ln(), inner is x^2 + 1:
g = sp.log(x**2 + 1)
dg_dx = sp.diff(g, x)
print("Composite Function (Chain Rule) Example:")
print(f"Function: g(x) = {g}")
print(f"Derivative: g'(x) = {dg_dx}\n")
```

```python
# Inverse Function Differentiation: Let's find the derivative of the inverse of h(x) = e^x, which is h^-1(x) = ln(x):
h = sp.exp(x)
h_inverse = sp.ln(x)
dh_inverse_dx = sp.diff(h_inverse, x)
print("Inverse Function Example:")
print(f"Original Function: h(x) = {h}")
print(f"Inverse Function:  h^-1(x) = {h_inverse}")
print(f"Derivative of Inverse: [h^-1(x)]' = {dh_inverse_dx}")
```
