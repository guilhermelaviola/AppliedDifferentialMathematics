# **Continuity & Intermediate Value Theorem**
**Continuity** and the **Intermediate Value Theorem (IVT)** are foundational calculus concepts describing the unbroken behavior of functions. A function is continuous over an interval if its graph can be drawn without lifting your pencil from the paper, meaning it has no holes, jumps, or vertical asymptotes, and its limit at any point equals its actual value. The Intermediate Value Theorem directly builds on this property: it states that if a function is continuous on a closed interval $[a, b]$, it must take on every single value between $f(a)$ and $f(b)$ at least once. Essentially, because the function cannot 'jump' over any numbers, it is guaranteed to cross every intermediate height—a principle famously used to prove that a continuous function changing from negative to positive must have a root (or zero) somewhere in between.

## **Example: Intermediate Value Theorem**
In the following example we use a continuous function, $f(x) = x^3 - x - 2$, on the interval $[1, 2]$. Because $f(1) = -2$ (negative) and $f(2) = 4$ (positive), the IVT guarantees there is a root ($f(x) = 0$) between $1$ and $2$. We can find this root using the Bisection Method, which relies entirely on the IVT.

```python
# Defining the functions:
def f(x):
    # A continuous polynomial function: f(x) = x^3 - x - 2:
    return x**3 - x - 2

def bisection_method(a, b, tol=1e-6):
    # Verifying the Intermediate Value Theorem applies. The function must change signs over the interval:
    if f(a) * f(b) >= 0:
        print('IVT condition not met: The function must have opposite signs at the boundaries.')
        return None

    print(f'Starting interval: [{a}, {b}] where f({a}) = {f(a)} and f({b}) = {f(b)}')
    print('-' * 50)

    # Iteratively narrowing down the interval:
    while (b - a) / 2.0 > tol:
        midpoint = (a + b) / 2.0
        f_mid = f(midpoint)

        print(f'Current Midpoint: {midpoint:.6f} | f(midpoint) = {f_mid:.6f}')

        if abs(f_mid) < tol:  # Found a root close enough to tolerance
            return midpoint
        elif f(a) * f_mid < 0:  # The root is in the left half
            b = midpoint
        else:  # The root is in the right half
            a = midpoint

    return (a + b) / 2.0

# Defining the closed interval [a, b]:
interval_a = 1
interval_b = 2

# Running the algorithm:
root = bisection_method(interval_a, interval_b)

if root:
    print('-' * 50)
    print(f'By the Intermediate Value Theorem, a root exists!')
    print(f'Approximate root found at x = {root:.6f}')
    print(f'Verification: f({root:.6f}) = {f(root):.6f}')
```
