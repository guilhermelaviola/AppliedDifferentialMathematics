# **Extrema, Concavity & Inflection Points**
In calculus, **extrema** represent the absolute or relative maximum and minimum values of a function, occurring where the first derivative is zero ($f'(x) = 0$) or undefined, which signifies a peak or valley in the graph. **Concavity** describes the direction in which the curve bends, dictated by the second derivative ($f''(x)$): a function is concave up (shaped like a cup) where $f''(x) > 0$ and concave down (shaped like a frown) where $f''(x) < 0$. An **inflection point** is the precise location on the graph where this concavity changes from upward to downward (or vice versa), which typically occurs where the second derivative is zero ($f''(x) = 0$) or does not exist, provided the sign of $f''(x)$ actually flips.

## **Example: Extrema, Concavity & Inflection Points**
Following is an example using SymPy (a symbolic mathematics library) to automatically find the extrema, concavity, and inflection points of a sample cubic function: $f(x) = x^3 - 3x^2$

```python
# Importing all the necessary libraries:
import sympy as sp
```

```python
# Defining the variable and the function
x = sp.Symbol('x')
f = x**3 - 3*x**2
print(f'Function: f(x) = {f}\n')
```

```python
# Finding Extrema (First Derivative):
f_prime = sp.diff(f, x)
critical_points = sp.solve(f_prime, x)

print('Extrema Analysis')
print(f"First Derivative: f'(x) = {f_prime}")
print(f"Critical Points (where f'(x) = 0): {critical_points}")
```

```python
# Using the second derivative test to classify extrema:
for pt in critical_points:
    f_double_prime_val = sp.diff(f_prime, x).subs(x, pt)
    y_val = f.subs(x, pt)
    if f_double_prime_val < 0:
        print(f"  -> Local Maximum at ({pt}, {y_val}) [f''(x) < 0]")
    elif f_double_prime_val > 0:
        print(f"  -> Local Minimum at ({pt}, {y_val}) [f''(x) > 0]")
```

```python
# Finding Concavity & Inflection Points (Second Derivative):
f_double_prime = sp.diff(f, x, 2)
inflection_points = sp.solve(f_double_prime, x)

print('Concavity & Inflection Points')
print(f"Second Derivative: f''(x) = {f_double_prime}")
print(f"Potential Inflection Points (where f''(x) = 0): {inflection_points}")

for pt in inflection_points:
    y_val = f.subs(x, pt)
    print(f'  -> Inflection Point confirmed at ({pt}, {y_val})')
```

```python
# Checking concavity in intervals around the inflection point (x = 1):
concave_down_test = f_double_prime.subs(x, 0) # Test x < 1
concave_up_test = f_double_prime.subs(x, 2)   # Test x > 1

print(f'  -> Concavity for x < 1 (e.g., x=0): {'Concave Down' if concave_down_test < 0 else 'Concave Up'}')
print(f'  -> Concavity for x > 1 (e.g., x=2): {'Concave Down' if concave_up_test < 0 else 'Concave Up'}')
```
